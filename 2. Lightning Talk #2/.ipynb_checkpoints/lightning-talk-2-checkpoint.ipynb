### Outlier Decision
LIMIT_BAL, bill amounts, and payment amounts all show long right tails with many high-value outliers, but these are legitimate high-spending/high-limit customers, not data errors — none are negative or impossibly large.
Decision: outliers are kept as-is rather than removed, since removing them would discard real customer behavior. LIMIT_BAL and the bill/payment amount columns will instead be log-transformed during preprocessing to reduce skew for models sensitive to scale (Logistic Regression); tree-based models (Random Forest, Gradient Boosting) will use the raw values since they are not affected by skew or outliers.

```python
corr_cols = continuous_cols + ['default.payment.next.month']

correlation_matrix = df_cleaned[corr_cols].corr()

correlation_matrix
```

```python
plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    cmap='coolwarm',
    center=0
)

plt.title('Correlation Matrix of Continuous Variables')
plt.show()
```

Bill amount columns (BILL_AMT1–6) are very highly correlated with each other (>0.9), which is a multicollinearity risk to keep in mind for Logistic Regression — feature engineering (e.g., average bill amount) can help consolidate this.

```python
repay_target_corr = (
    df_cleaned[repay_cols + ['default.payment.next.month']]
    .corr()['default.payment.next.month']
    .sort_values(ascending=False)
)

repay_target_corr
```

PAY_0 (most recent repayment status) has the strongest correlation with default (0.32) of any single feature, and correlation strength decreases the further back in time you go (PAY_2 → PAY_6) — recent behavior matters most.

```python
categorical_cols = ['SEX', 'EDUCATION', 'MARRIAGE']

chi_square_results = []

for col in categorical_cols:
    contingency_table = pd.crosstab(
        df_cleaned[col],
        df_cleaned['default.payment.next.month']
    )
    
    chi2, p_value, dof, expected = chi2_contingency(contingency_table)
    
    chi_square_results.append({
        'Feature': col,
        'Chi2': chi2,
        'p_value': p_value
    })

chi_square_results = pd.DataFrame(chi_square_results)

chi_square_results
```

```python
categorical_cols = [
    'SEX', 'EDUCATION', 'MARRIAGE',
    'PAY_0', 'PAY_2', 'PAY_3',
    'PAY_4', 'PAY_5', 'PAY_6'
]

chi_square_results = []

for col in categorical_cols:
    contingency_table = pd.crosstab(
        df_cleaned[col],
        df_cleaned['default.payment.next.month']
    )

    chi2, p_value, dof, expected = chi2_contingency(contingency_table)

    chi_square_results.append({
        'Feature': col,
        'Chi2': chi2,
        'p_value': p_value
    })

chi_square_results = pd.DataFrame(chi_square_results)
chi_square_results.sort_values('Chi2', ascending=False)
```

All PAY_x columns are overwhelmingly significant (chi2 in the thousands) compared to demographic features (chi2 in the tens-to-hundreds), confirming repayment history should be weighted most heavily in modeling.

# Data Preprocessing & Feature Engineering
Transforming, encoding, standardization and preparing features
- **You only need to outline your approach here. No need to execute**

### Proposed Preprocessing

- Drop the `ID` column because it is only a unique identifier and does not provide predictive information.
- Keep `SEX`, `EDUCATION`, and `MARRIAGE` as categorical features.
- Keep `PAY_0` to `PAY_6` as ordinal repayment-status features.
- Standardize continuous numerical features such as `LIMIT_BAL`, `AGE`, bill amounts, and payment amounts for models that are sensitive to scale.
- Consider transforming highly skewed bill and payment amount variables.
- Use a stratified train-test split to preserve the class distribution.
- Consider class weights because the target is moderately imbalanced.

  ### Proposed Feature Engineering

Potential engineered features include:

- Number of delayed-payment months.
- Maximum repayment delay across the six months.
- Average bill amount.
- Average payment amount.
- Payment-to-bill ratio.
- Credit utilization ratio.
