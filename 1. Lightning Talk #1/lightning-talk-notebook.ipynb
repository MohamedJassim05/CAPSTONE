# Topic 1 : Loan Status Prediction

## Problem Statement

Can we predict the status of a loan using the borrower's financial information, employment details, loan characteristics, and repayment history?


## Data Info
Data Source : https://www.kaggle.com/datasets/datawitharyan/financial-loan-dataset/data

## Data Dictionary

| Column | Type | Description |
|--------|------|-------------|
| `id` | Integer | Unique identifier for each loan record. |
| `address_state` | Categorical | State where the borrower lives. |
| `application_type` | Categorical | Whether the loan application was individual or joint. |
| `emp_length` | Categorical | Length of the borrower's employment. |
| `emp_title` | Categorical | Borrower's job title. |
| `grade` | Categorical | Overall loan grade assigned by the lender. |
| `home_ownership` | Categorical | Home ownership status (e.g., RENT, OWN, MORTGAGE). |
| `issue_date` | Date | Date when the loan was issued. |
| `last_credit_pull_date` | Date | Most recent date the borrower's credit report was checked. |
| `last_payment_date` | Date | Date of the borrower's last loan payment. |
| `loan_status` | Categorical | **Target variable** indicating the current loan outcome (e.g., Fully Paid, Charged Off, Current). |
| `next_payment_date` | Date | Scheduled date of the next loan payment. |
| `purpose` | Categorical | Purpose of the loan (e.g., debt consolidation, home improvement). |
| `sub_grade` | Categorical | More detailed loan grade within each grade category. |
| `term` | Categorical | Loan repayment period (e.g., 36 months or 60 months). |
| `verification_status` | Categorical | Indicates whether the borrower's income was verified. |
| `member_id` | Integer | Unique identifier for the borrower. |
| `annual_income` | Numerical | Borrower's annual income. |
| `dti` | Numerical | Debt-to-income ratio. |
| `installment` | Numerical | Fixed monthly payment amount. |
| `int_rate` | Numerical | Interest rate charged on the loan. |
| `loan_amount` | Numerical | Total loan amount borrowed. |
| `total_acc` | Numerical | Total number of credit accounts held by the borrower. |
| `total_payment` | Numerical | Total amount paid toward the loan. |

## Target Variable

```python
[col for col in loan.columns if 'status' in col.lower()]
```

```python
loan['loan_status'].value_counts()
```

```python
loan['loan_status'].value_counts(normalize=True) * 100
```

```python
loan['loan_status'].value_counts().plot(kind='bar')

plt.title('Loan Status Distribution')
plt.xlabel('Loan Status')
plt.ylabel('Number of Loans')
plt.xticks(rotation=0)
plt.show()
```

The target column is `loan_status`. It represents the current outcome of each loan: Fully Paid, Charged Off, or Current. Since the target contains categories, this is a multiclass classification problem. The target is imbalanced because 83.33% of the loans are Fully Paid, while Charged Off and Current loans represent much smaller groups.

```python
categorical_columns = loan.select_dtypes(
    include=['object', 'string', 'category']
).columns.tolist()

numerical_columns = loan.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

print("Number of categorical columns:", len(categorical_columns))
print("Number of numerical columns:", len(numerical_columns))
```

```python
print("Categorical columns:")
print(categorical_columns)

print("\nNumerical columns:")
print(numerical_columns)
```

The dataset originally contains 15 categorical columns and 9 numerical columns. However, loan_status is the target, while id and member_id are identifiers rather than useful predictors.

## Correlation Analysis

```python
loan_corr = loan.copy()

loan_corr['loan_status_encoded'] = loan_corr['loan_status'].map({
    'Fully Paid': 0,
    'Current': 1,
    'Charged Off': 2
})
```

```python
target_correlation = (
    loan_corr.select_dtypes(include=['int64', 'float64'])
    .corr()['loan_status_encoded']
    .sort_values(ascending=False)
)

target_correlation
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

numeric_data = loan_corr.select_dtypes(include=['int64', 'float64'])

plt.figure(figsize=(12, 8))
sns.heatmap(
    numeric_data.corr(),
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')
plt.show()
```

int_rate had the strongest positive correlation with the encoded target at about 0.23, while total_payment had the strongest negative correlation at about -0.19. Strong correlations were also found between installment, loan_amount, and total_payment.

## Outlier Detection

```python
outlier_columns = [
    col for col in numerical_columns
    if col not in ['id', 'member_id']
]
```

```python
for column in outlier_columns:
    plt.figure(figsize=(8, 3))
    sns.boxplot(x=loan[column])
    plt.title(f'Boxplot of {column}')
    plt.show()
```

```python
for column in outlier_columns:
    Q1 = loan[column].quantile(0.25)
    Q3 = loan[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = loan[
        (loan[column] < lower_limit) |
        (loan[column] > upper_limit)
    ]

    print(column, ":", len(outliers), "outliers")
```

Outliers were found in annual_income, installment, int_rate, loan_amount, total_acc, and total_payment. I would not remove them automatically because they may be valid financial values. I would inspect them first and consider robust scaling, log transformation, or IQR capping if needed.

```python
for column in outlier_columns:
    Q1 = bankruptcy[column].quantile(0.25)
    Q3 = bankruptcy[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = bankruptcy[
        (bankruptcy[column] < lower_limit) |
        (bankruptcy[column] > upper_limit)
    ]

    print(column, ":", len(outliers), "outliers")
```

Outliers were found in most financial variables, especially net income, retained earnings, current assets, liabilities, and EBIT.
