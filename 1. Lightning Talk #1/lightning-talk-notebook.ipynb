# Topic 1 : Loan Status Prediction

## Problem Statement

Can we predict the status of a loan using the borrower's financial information, employment details, loan characteristics, and repayment history?


## Data Info
Data Source : https://www.kaggle.com/datasets/datawitharyan/financial-loan-dataset/data

## Data Dictionary

| Column | Type | Description |
|--------|------|-------------|
| `id` | Integer | Unique identifier for each loan record. |
| `address_state` | Categorical | State where the borrower lives. |
| `application_type` | Categorical | Whether the loan application was individual or joint. |
| `emp_length` | Categorical | Length of the borrower's employment. |
| `emp_title` | Categorical | Borrower's job title. |
| `grade` | Categorical | Overall loan grade assigned by the lender. |
| `home_ownership` | Categorical | Home ownership status (e.g., RENT, OWN, MORTGAGE). |
| `issue_date` | Date | Date when the loan was issued. |
| `last_credit_pull_date` | Date | Most recent date the borrower's credit report was checked. |
| `last_payment_date` | Date | Date of the borrower's last loan payment. |
| `loan_status` | Categorical | **Target variable** indicating the current loan outcome (e.g., Fully Paid, Charged Off, Current). |
| `next_payment_date` | Date | Scheduled date of the next loan payment. |
| `purpose` | Categorical | Purpose of the loan (e.g., debt consolidation, home improvement). |
| `sub_grade` | Categorical | More detailed loan grade within each grade category. |
| `term` | Categorical | Loan repayment period (e.g., 36 months or 60 months). |
| `verification_status` | Categorical | Indicates whether the borrower's income was verified. |
| `member_id` | Integer | Unique identifier for the borrower. |
| `annual_income` | Numerical | Borrower's annual income. |
| `dti` | Numerical | Debt-to-income ratio. |
| `installment` | Numerical | Fixed monthly payment amount. |
| `int_rate` | Numerical | Interest rate charged on the loan. |
| `loan_amount` | Numerical | Total loan amount borrowed. |
| `total_acc` | Numerical | Total number of credit accounts held by the borrower. |
| `total_payment` | Numerical | Total amount paid toward the loan. |

The target column is `loan_status`. It represents the current outcome of each loan: Fully Paid, Charged Off, or Current. ## Correlation Analysis

```python
loan_corr = loan.copy()

loan_corr['loan_status_encoded'] = loan_corr['loan_status'].map({
    'Fully Paid': 0,
    'Current': 1,
    'Charged Off': 2
})
```

```python
target_correlation = (
    loan_corr.select_dtypes(include=['int64', 'float64'])
    .corr()['loan_status_encoded']
    .sort_values(ascending=False)
)

target_correlation
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

numeric_data = loan_corr.select_dtypes(include=['int64', 'float64'])

plt.figure(figsize=(12, 8))
sns.heatmap(
    numeric_data.corr(),
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')
plt.show()
```

int_rate had the strongest positive correlation with the encoded target at about 0.23, while total_payment had the strongest negative correlation at about -0.19. Strong correlations were also found between installment, loan_amount, and total_payment.

## Outlier Detection

```python
outlier_columns = ['annual_income', 'installment', 'int_rate', 'loan_amount', 'total_acc', 'total_payment']

outlier_counts = {}
for column in outlier_columns:
    Q1 = loan[column].quantile(0.25)
    Q3 = loan[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = loan[(loan[column] < lower_limit) | (loan[column] > upper_limit)]
    outlier_counts[column] = len(outliers)
    print(column, ":", len(outliers), "outliers")
```

```python
worst_columns = sorted(outlier_counts, key=outlier_counts.get, reverse=True)[:2]

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for ax, column in zip(axes, worst_columns):
    sns.boxplot(x=loan[column], ax=ax)
    ax.set_title(f'Distribution of {column}')
plt.tight_layout()
plt.show()

for column in worst_columns:
    print(f"\nTop 10 highest {column}:")
    print(loan[column].nlargest(10))
```

```python
for column in outlier_columns:
    suspicious = loan[column].isin([9999, 99999, 999999])
    if suspicious.sum() > 0:
        print(column, ":", suspicious.sum(), "rows with suspicious round-number values")
```

annual_income and total_payment are right-skewed and contain upper-end outliers. The highest annual income is approximately 6,000,000, while the highest total payment is about 58,564. I also checked for suspicious round-number values and found only a few cases, which are not enough to suggest a data quality issue. These values may represent genuine borrowers with high income or large repayments

## Outlier Analysis

retained_earnings and net_income contain both upper-end and lower-end outliers. The highest retained earnings value is approximately 402,089, while the highest net income is about 104,821. Negative extreme values also appear, which may represent companies experiencing major losses. The round-number check found only a small number of suspicious values, so there is no strong sign of a data-quality problem.

# Topic 3 : Credit Card Default Prediction

## Problem Statement

```python
outlier_columns = [col for col in numerical_columns if col != 'ID']

outlier_counts = {}
for column in outlier_columns:
    Q1 = credit[column].quantile(0.25)
    Q3 = credit[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = credit[(credit[column] < lower_limit) | (credit[column] > upper_limit)]
    outlier_counts[column] = len(outliers)
    print(column, ":", len(outliers), "outliers")
```

Outliers were mainly found in BILL_AMT1 and PAY_AMT1. The highest bill amount is approximately 964,511, while the highest payment amount is about 873,552. These extreme values may represent customers with unusually large bills or payments rather than data errors.
